```python
# AB Mean Plant Height (cm) (Hacker, 2019)
H_MHeight_AB = Hacker_2019["Mean Plant Height (cm)"][0]
print(H_MHeight_AB)

# AB 95CI Hacker
H_AB_95CI = Hacker_2019["Mean Plant Height (± 95% CI)"][0]
print(H_AB_95CI)

# SP Mean Plant Height (cm) (Hacker, 2019)
H_MHeight_SP = Hacker_2019["Mean Plant Height (cm)"][1]
print(H_MHeight_SP)

# SP 95CI Hacker
H_SP_95CI = Hacker_2019["Mean Plant Height (± 95% CI)"][1]
print(H_SP_95CI)

# UP Mean Plant Height (cm) (Hacker, 2019)
H_MHeight_UP = Hacker_2019["Mean Plant Height (cm)"][2]
print(H_MHeight_UP)

# UP 95CI Hacker
H_UP_95CI = Hacker_2019["Mean Plant Height (± 95% CI)"][2]
print(H_UP_95CI)
```

```python

import numpy as np
import matplotlib.pyplot as plt

# Species
species = ["Ammophila", "Spartina", "Uniola"]

# Mean plant heights (cm)

riffe_3year = [R_MHeight_AB, R_MHeight_SP, R_MHeight_UP]

riffe_preplanting = [
    R_MPre_Height_AB,
    R_MPre_Height_SP,
    R_MPre_Height_UP
]

carothers = [
    C_MHeight_AB,
    C_MHeight_SP,
    C_MHeight_UP
]

hacker = [
    H_MHeight_AB,  # Ammophila
    H_MHeight_SP,  # Spartina
    Hacker_2019["Mean Plant Height (cm)"][2]  # Uniola
]

# 95% confidence intervals
riffe_3year_ci = [
    R_MHeight_95CI_AB,
    R_MHeight_95CI_SP,
    R_MHeight_95CI_UP
]

riffe_preplanting_ci = [
    R_MPre_Height_95CI_AB,
    R_MPre_Height_95CI_SP,
    R_MPre_Height_95CI_UP
]

carothers_ci = [
    C_AB_95CI,
    C_SP_95CI,
    C_UP_95CI
]

hacker_ci = [
    H_AB_95CI,
    H_SP_95CI,
    Hacker_2019["Mean Plant Height (± 95% CI)"][2]
]

# Set up bar chart
x = np.arange(len(species))
width = 0.2

fig, ax = plt.subplots(figsize=(12, 7))

# Four study groups


hacker_height_bars = ax.bar(x - 1.5*width, hacker, width,
       yerr=hacker_ci, capsize=4, label="Hacker (2019)")

riffe_preplanting_bars = ax.bar(x - 0.5*width, riffe_preplanting, width,
       yerr=riffe_preplanting_ci, capsize=4,
       label="Riffe Pre-Planting (2021)")

riffe_3_year_bars =ax.bar(x + 0.5*width, riffe_3year, width,
       yerr=riffe_3year_ci, capsize=4, label="Riffe 3-Year (2021-2023)")

carothers_height_bars = ax.bar(x + 1.5*width, carothers, width,
       yerr=carothers_ci, capsize=4, label="Carothers (2026)", color='red')

# Labels and formatting
ax.set_xlabel("Plant Species", fontsize=12)
ax.set_ylabel("Mean Plant Height (cm)", fontsize=12)
ax.set_title("Mean Plant Height Across Four Studies", fontsize=15)
ax.bar_label(carothers_height_bars, padding=3, fontsize=10, fmt='%.2f')
ax.bar_label(hacker_height_bars, padding=3, fontsize=10, fmt='%.2f')
ax.bar_label(riffe_3_year_bars, padding=3, fontsize=10, fmt='%.2f')
ax.bar_label(riffe_preplanting_bars, padding=3, fontsize=10, fmt='%.2f')

ax.set_xticks(x)
ax.set_xticklabels(species)

ax.legend(title="Study")
ax.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```
